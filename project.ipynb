```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv('huge_1M_titanic.csv')
data
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data=data.drop(columns=['PassengerId','Name','Age','Ticket','Cabin'])
```

```python
data.head()
```

```python
data['Embarked'].value_counts()
```

```python
data['Embarked']=data['Embarked'].replace({'S':'Southampton','C':'Chebourge','Q':'Queenstown'})
```

```python
data['Embarked'].value_counts()
```

```python
data.isnull().sum()
```

```python
data.dropna(subset=['Embarked'],inplace=True)
```

```python
data.isnull().sum()
```

```python
data
```

```python
data.shape
```

```python
from sklearn.preprocessing import LabelEncoder
label=LabelEncoder()
```

```python
data['Sex']=label.fit_transform(data['Sex'])
```

```python
from sklearn.preprocessing import OneHotEncoder
onehot=OneHotEncoder(sparse_output=False)
```

```python
Embarked = onehot.fit_transform(data[['Embarked']])
```

```python
Embarked=pd.DataFrame(Embarked,columns=onehot.get_feature_names_out())
```

```python
Embarked
```

```python
data=pd.concat([data.drop(columns=['Embarked']),Embarked],axis=1)
```

```python
data.head()
```

```python
from sklearn.preprocessing import StandardScaler
scale= StandardScaler()
```

```python
data.columns
```

```python
num_columns= ['Pclass', 'SibSp', 'Parch', 'Fare']
```

```python
data[num_columns]=scale.fit_transform(data[num_columns])
```

```python
X=data.drop(columns='Survived')
```

```python
y=data['Survived']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=20,random_state=42)
```

```python
X_train,X_valid,y_train,y_valid=train_test_split(X_train,y_train,test_size=0.2,random_state=42)
```

```python
import tensorflow
```

```python
X_train.shape
```

```python
X_valid.shape
```

model building

```python
(X_train.shape[1],)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Input
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model=Sequential([
 Input(shape=(X_train.shape[1],)),
 Dense(64,activation='relu'),
 Dense(32,activation='relu'),
 Dense(1,activation='sigmoid')   
])
```

```python
model.summary()
```

```python
opt=tensorflow.keras.optimizers.Adam(learning_rate=0.01)
tensorflow.keras.losses.BinaryFocalCrossentropy
```

```python
model.compile(optimizer=opt,loss='binary_crossentropy',metrics=['accuracy'])
```

```python
stoping_callback=EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
```

```python
model.fit(X_train,y_train,validation_data=(X_valid,y_valid),epochs=10,callbacks=[stoping_callback])
```

```python
model.save('model.h5')
```
